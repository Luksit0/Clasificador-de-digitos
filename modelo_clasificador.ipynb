```python
# Librerias necesarias.
import numpy as np
import tensorflow as tf
import matplotlib.pyplot as plt
from tensorflow.keras import layers
```

## Parte 1. Preparación de los datos y Dataset

```python
# Cargar el dataset
(x_train,y_train), (x_test, y_test) = tf.keras.datasets.mnist.load_data()

# Normalixación de los pixeles

x_train = (x_train.astype("float32")/255.0)[..., np.newaxis]
x_test = (x_train.astype("float32")/255.0)[..., np.newaxis]

# Visualización de la forma correcta para el CNN
print("x_train", x_train.shape, x_train.dtype, "min/max", x_train.min(), x_train.max())
print("y_train", y_train.shape, y_train.dtype, "clases", np.unique(y_train))
print("x_test", x_test.shape, x_test.dtype, "min/max", x_test.min(), x_test.max())
print("y_test", y_test.shape, y_test.dtype)
```

### Visualización de las 5 imagenes

```python
fig, axes = plt.subplots(3,5, figsize= (10,4))
for i, ax in enumerate(axes.flat):
    ax.imshow(x_train[i, :, :, 0], cmap="Greens", vmin=0.0, vmax=1.0)
    ax.set_title(f"Etiqueta: {y_train[i]}")
    ax.axis("off")

plt.tight_layout()
plt.show()
```
